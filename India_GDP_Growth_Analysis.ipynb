```python
# Recreate India GDP Growth dataset

import pandas as pd
import requests

# Load World Bank GDP Growth file
df = pd.read_excel(
    "GDP_DataSet.xls",
    sheet_name="Data",
    header=0
)

# Select India
india_df = df[df["Country Code"] == "IND"].copy()

# Select 2000–2025
years = [str(year) for year in range(2000, 2026)]

india_gdp = india_df[years].T.reset_index()

india_gdp.columns = ["Year", "GDP_Growth"]

india_gdp["Year"] = india_gdp["Year"].astype(int)

india_gdp["GDP_Growth"] = pd.to_numeric(
    india_gdp["GDP_Growth"],
    errors="coerce"
)

# Create 5-year moving average
india_gdp["5Y_Moving_Average"] = (
    india_gdp["GDP_Growth"].rolling(5).mean()
)

# Create decade
india_gdp["Decade"] = (india_gdp["Year"] // 10) * 10

india_gdp
```

```python
# Create master economic dataset

# 1. Inflation
url = "https://api.worldbank.org/v2/country/IND/indicator/FP.CPI.TOTL.ZG"

params = {
    "format": "json",
    "date": "2000:2025",
    "per_page": 100
}

response = requests.get(url, params=params)

inflation_df = pd.DataFrame(response.json()[1])
inflation_df = inflation_df[["date", "value"]]
inflation_df.columns = ["Year", "Inflation"]

inflation_df["Year"] = inflation_df["Year"].astype(int)
inflation_df["Inflation"] = pd.to_numeric(
    inflation_df["Inflation"], errors="coerce"
)


# 2. Population
url = "https://api.worldbank.org/v2/country/IND/indicator/SP.POP.TOTL"

response = requests.get(url, params=params)

population_df = pd.DataFrame(response.json()[1])
population_df = population_df[["date", "value"]]
population_df.columns = ["Year", "Population"]

population_df["Year"] = population_df["Year"].astype(int)
population_df["Population"] = pd.to_numeric(
    population_df["Population"], errors="coerce"
)


# 3. GDP Per Capita
url = "https://api.worldbank.org/v2/country/IND/indicator/NY.GDP.PCAP.CD"

response = requests.get(url, params=params)

gdp_pc_df = pd.DataFrame(response.json()[1])
gdp_pc_df = gdp_pc_df[["date", "value"]]
gdp_pc_df.columns = ["Year", "GDP_Per_Capita"]

gdp_pc_df["Year"] = gdp_pc_df["Year"].astype(int)
gdp_pc_df["GDP_Per_Capita"] = pd.to_numeric(
    gdp_pc_df["GDP_Per_Capita"], errors="coerce"
)


# 4. Merge everything
india_economic = india_gdp.merge(
    inflation_df,
    on="Year",
    how="inner"
).merge(
    population_df,
    on="Year",
    how="inner"
).merge(
    gdp_pc_df,
    on="Year",
    how="inner"
)


# 5. Population Growth
india_economic["Population_Growth"] = (
    india_economic["Population"].pct_change() * 100
)

# 6. GDP Per Capita Growth
india_economic["GDP_Per_Capita_Growth"] = (
    india_economic["GDP_Per_Capita"].pct_change() * 100
)

india_economic
```

```python
import requests
import pandas as pd

# -----------------------------
# 1. Inflation Data
# -----------------------------
url_inflation = "https://api.worldbank.org/v2/country/IND/indicator/FP.CPI.TOTL.ZG"

params = {
    "format": "json",
    "date": "1995:2025",
    "per_page": 100
}

response = requests.get(url_inflation, params=params)
inflation_data = response.json()[1]

inflation_df = pd.DataFrame(inflation_data)
inflation_df = inflation_df[["date", "value"]]
inflation_df.columns = ["Year", "Inflation"]
inflation_df["Year"] = inflation_df["Year"].astype(int)
inflation_df["Inflation"] = pd.to_numeric(
    inflation_df["Inflation"], errors="coerce"
)


# -----------------------------
# 2. Population Data
# -----------------------------
url_population = "https://api.worldbank.org/v2/country/IND/indicator/SP.POP.TOTL"

response = requests.get(url_population, params=params)
population_data = response.json()[1]

population_df = pd.DataFrame(population_data)
population_df = population_df[["date", "value"]]
population_df.columns = ["Year", "Population"]
population_df["Year"] = population_df["Year"].astype(int)
population_df["Population"] = pd.to_numeric(
    population_df["Population"], errors="coerce"
)


# -----------------------------
# 3. GDP Per Capita Data
# -----------------------------
url_gdp_pc = "https://api.worldbank.org/v2/country/IND/indicator/NY.GDP.PCAP.CD"

response = requests.get(url_gdp_pc, params=params)
gdp_pc_data = response.json()[1]

gdp_pc_df = pd.DataFrame(gdp_pc_data)
gdp_pc_df = gdp_pc_df[["date", "value"]]
gdp_pc_df.columns = ["Year", "GDP_Per_Capita"]
gdp_pc_df["Year"] = gdp_pc_df["Year"].astype(int)
gdp_pc_df["GDP_Per_Capita"] = pd.to_numeric(
    gdp_pc_df["GDP_Per_Capita"], errors="coerce"
)


# -----------------------------
# 4. Create GDP Growth Dataset
# -----------------------------
gdp = india_gdp.iloc[0]

gdp_growth_df = pd.DataFrame({
    "Year": list(range(1960, 2026)),
    "GDP_Growth": pd.to_numeric(
        gdp.iloc[3:].values,
        errors="coerce"
    )
})

gdp_growth_df["Moving_Average_5Y"] = (
    gdp_growth_df["GDP_Growth"]
    .rolling(5)
    .mean()
)

gdp_growth_df["Decade"] = (
    gdp_growth_df["Year"] // 10
) * 10


# -----------------------------
# 5. Population Growth
# -----------------------------
population_df = population_df.sort_values("Year")

population_df["Population_Growth"] = (
    population_df["Population"]
    .pct_change() * 100
)


# -----------------------------
# 6. GDP Per Capita Growth
# -----------------------------
gdp_pc_df = gdp_pc_df.sort_values("Year")

gdp_pc_df["GDP_Per_Capita_Growth"] = (
    gdp_pc_df["GDP_Per_Capita"]
    .pct_change() * 100
)


# -----------------------------
# 7. Merge Everything
# -----------------------------
complete_economic = gdp_growth_df.merge(
    inflation_df,
    on="Year",
    how="inner"
)

complete_economic = complete_economic.merge(
    population_df,
    on="Year",
    how="inner"
)

complete_economic = complete_economic.merge(
    gdp_pc_df,
    on="Year",
    how="inner"
)


# -----------------------------
# 8. Get Missing 2000–2003
# -----------------------------
missing_full = complete_economic[
    complete_economic["Year"].between(2000, 2003)
].copy()

missing_full
```
